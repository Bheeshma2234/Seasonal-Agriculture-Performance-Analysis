```python
# ==========================================
# CELL 35: PROFITABILITY BY SEASON
# ==========================================

# ------------------------------------------------
# STEP 1: Find the actual profit column
# ------------------------------------------------

profit_candidates = [
    col for col in df_clean.columns
    if 'profit' in col.lower()
]

print("Profit-related columns found:")
print(profit_candidates)


# ------------------------------------------------
# STEP 2: Select profit column
# ------------------------------------------------

if len(profit_candidates) > 0:

    profit_col = profit_candidates[0]

elif 'Calculated_Profit' in df_clean.columns:

    profit_col = 'Calculated_Profit'

else:

    print("\n❌ Profit column not found.")
    print("\nAvailable columns:")
    print(df_clean.columns.tolist())

    raise ValueError("No profit column available.")


print("\nUsing profit column:", profit_col)


# ------------------------------------------------
# STEP 3: Convert profit to numeric
# ------------------------------------------------

df_clean[profit_col] = pd.to_numeric(
    df_clean[profit_col],
    errors='coerce'
)


# ------------------------------------------------
# STEP 4: Create Profitability Status
# ------------------------------------------------

df_clean['Profitability_Status'] = df_clean[profit_col].apply(
    lambda x: 'Profitable' if x > 0
    else 'Loss' if x < 0
    else 'Break-even'
)


# ------------------------------------------------
# STEP 5: Count profitability status by season
# ------------------------------------------------

profit_status = pd.crosstab(
    df_clean['Season'],
    df_clean['Profitability_Status']
)


print("\nProfitability Status by Season:")
display(profit_status)


# ------------------------------------------------
# STEP 6: Convert counts to percentages
# ------------------------------------------------

profit_status_percentage = (
    profit_status
    .div(profit_status.sum(axis=1), axis=0)
    * 100
)

print("\nProfitability Percentage by Season:")
display(profit_status_percentage.round(2))


# ------------------------------------------------
# STEP 7: Stacked Bar Chart
# ------------------------------------------------

profit_status_percentage.plot(
    kind='bar',
    stacked=True,
    figsize=(9, 6)
)

plt.title("Profitability Status Across Seasons")
plt.xlabel("Season")
plt.ylabel("Percentage of Farms (%)")
plt.ylim(0, 100)
plt.legend(title="Profitability Status")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 41: PROFIT OUTLIER ANALYSIS
# ==========================================

# Find the actual profit column
profit_candidates = [
    col for col in df_clean.columns
    if 'profit' in col.lower()
]

# Select the profit column
if len(profit_candidates) > 0:
    profit_col = profit_candidates[0]

elif 'Calculated_Profit' in df_clean.columns:
    profit_col = 'Calculated_Profit'

else:
    print("❌ Profit column not found.")
    print("\nAvailable columns:")
    print(df_clean.columns.tolist())
    raise ValueError("No profit column available.")

print("Using profit column:", profit_col)


# Convert profit to numeric
df_clean[profit_col] = pd.to_numeric(
    df_clean[profit_col],
    errors='coerce'
)


# ==========================================
# BOX PLOT
# ==========================================

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df_clean,
    x='Season',
    y=profit_col
)

plt.title("Profit Distribution and Outliers Across Seasons")
plt.xlabel("Season")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()


# ==========================================
# OUTLIER DETECTION USING IQR
# ==========================================

print("\n==========================================")
print("PROFIT OUTLIERS BY SEASON")
print("==========================================")


def find_outliers(group):

    Q1 = group[profit_col].quantile(0.25)
    Q3 = group[profit_col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = group[
        (group[profit_col] < lower_limit) |
        (group[profit_col] > upper_limit)
    ]

    return len(outliers)


outlier_counts = (
    df_clean
    .groupby('Season')
    .apply(find_outliers, include_groups=False)
)

print("\nNumber of profit outliers in each season:")
display(outlier_counts)
```

```python
# ==========================================
# CELL 45: PESTICIDE VS YIELD
# ==========================================

# Find pesticide-related columns
pesticide_columns = [
    col for col in df_clean.columns
    if any(word in col.lower() for word in [
        'pesticide',
        'pest'
    ])
]

print("Pesticide-related columns found:")
print(pesticide_columns)


# ==========================================
# CHECK PESTICIDE COLUMN
# ==========================================

if len(pesticide_columns) == 0:

    print("\n❌ No pesticide column found.")

    print("\nAvailable columns in your dataset:")
    for col in df_clean.columns:
        print("-", col)

else:

    pesticide_col = pesticide_columns[0]

    print("\nUsing pesticide column:")
    print(pesticide_col)


    # Convert pesticide and yield columns to numeric
    df_clean[pesticide_col] = pd.to_numeric(
        df_clean[pesticide_col],
        errors='coerce'
    )

    df_clean['Yield_Tonnes_Ha'] = pd.to_numeric(
        df_clean['Yield_Tonnes_Ha'],
        errors='coerce'
    )


    # ==========================================
    # SCATTER PLOT
    # ==========================================

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df_clean,
        x=pesticide_col,
        y='Yield_Tonnes_Ha',
        hue='Season',
        alpha=0.7
    )

    plt.title(
        "Relationship Between Pesticide Usage and Crop Yield"
    )

    plt.xlabel("Pesticide Used")
    plt.ylabel("Yield (tonnes/ha)")

    plt.tight_layout()
    plt.show()


    # ==========================================
    # CORRELATION
    # ==========================================

    correlation = df_clean[
        [pesticide_col, 'Yield_Tonnes_Ha']
    ].corr().iloc[0, 1]

    print("\nCorrelation between Pesticide Usage and Yield:")
    print(round(correlation, 3))
```

```python
# ==========================================
# CELL 46: SEASONAL PERFORMANCE SUMMARY
# ==========================================

print("Creating Seasonal Performance Summary...")

# Make sure important numeric columns are numeric
numeric_columns = [
    'Yield_Tonnes_Ha',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3'
]

for col in numeric_columns:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')


# Find available economic columns safely
profit_columns = [
    col for col in df_clean.columns
    if 'profit' in col.lower()
]

revenue_columns = [
    col for col in df_clean.columns
    if 'revenue' in col.lower()
]

cost_columns = [
    col for col in df_clean.columns
    if 'cost' in col.lower()
]


# Create profit if it does not already exist
if len(profit_columns) > 0:

    profit_col = profit_columns[0]
    df_clean[profit_col] = pd.to_numeric(
        df_clean[profit_col],
        errors='coerce'
    )

elif len(revenue_columns) > 0 and len(cost_columns) > 0:

    revenue_col = revenue_columns[0]
    cost_col = cost_columns[0]

    df_clean[revenue_col] = pd.to_numeric(
        df_clean[revenue_col],
        errors='coerce'
    )

    df_clean[cost_col] = pd.to_numeric(
        df_clean[cost_col],
        errors='coerce'
    )

    df_clean['Calculated_Profit'] = (
        df_clean[revenue_col] - df_clean[cost_col]
    )

    profit_col = 'Calculated_Profit'

else:
    profit_col = None


# Create seasonal summary using only columns that actually exist
summary_columns = []

if 'Yield_Tonnes_Ha' in df_clean.columns:
    summary_columns.append('Yield_Tonnes_Ha')

if 'Water_Used_m3' in df_clean.columns:
    summary_columns.append('Water_Used_m3')

if 'Water_Efficiency_t_per_1000m3' in df_clean.columns:
    summary_columns.append('Water_Efficiency_t_per_1000m3')

if profit_col is not None:
    summary_columns.append(profit_col)


season_summary = df_clean.groupby('Season')[summary_columns].mean().round(2)

print("\nSeasonal Performance Summary:")
display(season_summary)
```

```python
# ==========================================
# CELL 47: SEASONAL YIELD COMPARISON
# ==========================================

season_yield = (
    df_clean.groupby('Season')['Yield_Tonnes_Ha']
    .mean()
    .sort_values(ascending=False)
)

print("Average Yield by Season:")
display(season_yield.round(2))

plt.figure(figsize=(9, 6))

sns.barplot(
    x=season_yield.index,
    y=season_yield.values
)

plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (tonnes/ha)")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 50: SEASONAL PROFIT ANALYSIS
# ==========================================

# Find profit column
profit_columns = [
    col for col in df_clean.columns
    if 'profit' in col.lower()
]

revenue_columns = [
    col for col in df_clean.columns
    if 'revenue' in col.lower()
]

cost_columns = [
    col for col in df_clean.columns
    if 'cost' in col.lower()
]


# Use existing profit column
if len(profit_columns) > 0:

    profit_col = profit_columns[0]

# Otherwise calculate profit
elif len(revenue_columns) > 0 and len(cost_columns) > 0:

    revenue_col = revenue_columns[0]
    cost_col = cost_columns[0]

    df_clean[revenue_col] = pd.to_numeric(
        df_clean[revenue_col],
        errors='coerce'
    )

    df_clean[cost_col] = pd.to_numeric(
        df_clean[cost_col],
        errors='coerce'
    )

    df_clean['Calculated_Profit'] = (
        df_clean[revenue_col] - df_clean[cost_col]
    )

    profit_col = 'Calculated_Profit'

else:

    profit_col = None


if profit_col is not None:

    df_clean[profit_col] = pd.to_numeric(
        df_clean[profit_col],
        errors='coerce'
    )

    season_profit = (
        df_clean.groupby('Season')[profit_col]
        .mean()
        .sort_values(ascending=False)
    )

    print("Average Profit by Season:")
    display(season_profit.round(2))

    plt.figure(figsize=(9, 6))

    sns.barplot(
        x=season_profit.index,
        y=season_profit.values
    )

    plt.title("Average Profit by Season")
    plt.xlabel("Season")
    plt.ylabel("Average Profit")
    plt.tight_layout()
    plt.show()

else:

    print("No profit/revenue/cost columns were found.")
```

```python
# ==========================================
# CELL 51: PROFITABILITY STATUS
# ==========================================

if profit_col is not None:

    df_clean['Profitability_Status'] = df_clean[profit_col].apply(
        lambda x:
        'Profitable' if x > 0
        else 'Loss' if x < 0
        else 'Break-even'
    )

    profitability = pd.crosstab(
        df_clean['Season'],
        df_clean['Profitability_Status']
    )

    print("Profitability Status by Season:")
    display(profitability)

    profitability_percentage = (
        profitability
        .div(profitability.sum(axis=1), axis=0)
        * 100
    ).round(2)

    print("\nProfitability Percentage by Season:")
    display(profitability_percentage)

    profitability_percentage.plot(
        kind='bar',
        stacked=True,
        figsize=(10, 6)
    )

    plt.title("Profitability Status by Season")
    plt.xlabel("Season")
    plt.ylabel("Percentage (%)")
    plt.legend(title="Status")
    plt.tight_layout()
    plt.show()

else:

    print("Profit column is not available.")
```

```python
# ==========================================
# CELL 52: CROP-WISE PERFORMANCE
# ==========================================

if 'Crop' in df_clean.columns:

    crop_yield = (
        df_clean.groupby('Crop')['Yield_Tonnes_Ha']
        .mean()
        .sort_values(ascending=False)
    )

    print("Average Yield by Crop:")
    display(crop_yield.round(2))

    plt.figure(figsize=(12, 6))

    sns.barplot(
        x=crop_yield.index,
        y=crop_yield.values
    )

    plt.title("Average Yield by Crop")
    plt.xlabel("Crop")
    plt.ylabel("Yield (tonnes/ha)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

else:

    print("Crop column not found.")
```

```python
# ==========================================
# CELL 53: IRRIGATION PERFORMANCE
# ==========================================

if 'Irrigation_Method' in df_clean.columns:

    irrigation_yield = (
        df_clean.groupby('Irrigation_Method')['Yield_Tonnes_Ha']
        .mean()
        .sort_values(ascending=False)
    )

    print("Average Yield by Irrigation Method:")
    display(irrigation_yield.round(2))

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=irrigation_yield.index,
        y=irrigation_yield.values
    )

    plt.title("Average Yield by Irrigation Method")
    plt.xlabel("Irrigation Method")
    plt.ylabel("Yield (tonnes/ha)")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

else:

    print("Irrigation_Method column not found.")
```

```python
# ==========================================
# CELL 54: SEASON AND IRRIGATION ANALYSIS
# ==========================================

if 'Irrigation_Method' in df_clean.columns:

    season_irrigation = pd.pivot_table(
        df_clean,
        values='Yield_Tonnes_Ha',
        index='Season',
        columns='Irrigation_Method',
        aggfunc='mean'
    ).round(2)

    print("Average Yield by Season and Irrigation Method:")
    display(season_irrigation)

    plt.figure(figsize=(10, 6))

    sns.heatmap(
        season_irrigation,
        annot=True,
        fmt='.2f'
    )

    plt.title("Yield by Season and Irrigation Method")
    plt.xlabel("Irrigation Method")
    plt.ylabel("Season")
    plt.tight_layout()
    plt.show()

else:

    print("Irrigation_Method column not found.")
```

```python
# ==========================================
# CELL 55: CORRELATION ANALYSIS
# ==========================================

# Select only numeric columns
numeric_df = df_clean.select_dtypes(include='number')

print("Numeric columns used for correlation:")
print(numeric_df.columns.tolist())

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

plt.title("Correlation Matrix of Agricultural Performance Variables")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 58: BEST SEASON
# ==========================================

season_performance = (
    df_clean.groupby('Season')
    .agg(
        Average_Yield=('Yield_Tonnes_Ha', 'mean'),
        Average_Water=('Water_Used_m3', 'mean'),
        Average_Water_Efficiency=(
            'Water_Efficiency_t_per_1000m3',
            'mean'
        )
    )
    .round(2)
)

print("Seasonal Performance:")
display(season_performance)

best_yield_season = season_performance[
    'Average_Yield'
].idxmax()

best_efficiency_season = season_performance[
    'Average_Water_Efficiency'
].idxmax()

print("\nBest Season for Yield:")
print(best_yield_season)

print("\nBest Season for Water Efficiency:")
print(best_efficiency_season)
```
